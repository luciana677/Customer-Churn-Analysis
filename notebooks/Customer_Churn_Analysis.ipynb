### Our business problem
What factors are associated with customer churn, and which customer segments should the company prioritize for retention?

TotalCharges is suspicious, We previously saw: TotalCharges = object even though it represents a numerical amount. And now we know:
df.isna().sum() → TotalCharges = 0

This tells us that there aren't actual NaN values, but it doesn't prove that every TotalCharges value is valid numerical data. The column could contain blank spaces let's test that:

```python
df["TotalCharges"].unique()
```

```python
pd.to_numeric(df["TotalCharges"], errors="coerce").isna().sum()
```

np.int64(11) means there are 11 values in TotalCharges that cannot be interpreted as numbers. And this explains why:
df.isna().sum() gave: TotalCharges    0

```python
#lets investigate the 11 values
df[pd.to_numeric(df["TotalCharges"], errors="coerce").isna()][
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges"]
]
```

All 11 problematic records have tenure = 0.
And their MonthlyCharges are valid: These are customers who have just joined, so they have no accumulated charges yet.

Therefore, for our analysis,
A blank TotalCharges for a customer with tenure = 0 can reasonably be treated as 0

### Clean TotalCharges

```python
#Convert the column to numeric
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
#Replace the 11 invalid values with 0
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

### Contract and churn
To investigate customer commitment.

```python
contract_churn=pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

contract_churn
```

Month-to-month customers are substantially more likely to churn, while two-year contract customers have the lowest churn rate. it suggests contract commitment is strongly associated with customer retention.

```python
#To calclate the exact gap beteen the min and max churn
contract_churn["Yes"].max() - contract_churn["Yes"].min()
```

Month-to-month customers have the highest churn rate, while two-year contract customers have the lowest, with a 39.88 percentage-point gap between them. 

Insight: Customers on longer-term contracts show substantially lower churn rates than month-to-month customers

### Tenure and churn

To investigate the customer lifecycle. Do churned customers actually have a lower average tenure than customers who stayed?

```python
df.groupby("Churn")["tenure"].mean()
```

Insight #3 — Tenure and churn

Customers who churned had an average tenure of 17.98 months, compared with 37.57 months for customers who stayed — a difference of approximately 19.59 months.

```python
df["tenure_group"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["0–12 months", "13–24 months", "25–48 months", "49–72 months"]
)
```

```python
tenure_churn = pd.crosstab(
    df["tenure_group"],
    df["Churn"],
    normalize="index"
) * 100

tenure_churn
```

Insight #4 — Churn is concentrated among newer customers

Customers in their first 12 months have the highest churn rate, while customers with 49–72 months of tenure have the lowest. Churn declines as customer tenure increases.

The first year is the critical period for customer retention.

```python
### Payment Method vs churn
```

```python
payment_churn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

payment_churn.round(2)
```

Insight #5 — Payment method and churn

Electronic check customers have the highest churn rate at 45.29%, while customers using automatic credit-card payments have the lowest at 15.24% — a 30.05 percentage point gap. 

Customers using electronic checks are substantially more likely to churn than customers using automatic credit-card payments.

```python
### Internet Service VS Churn
```

```python
internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

internet_churn.round(2)
```

Fiber-optic customers have the highest churn rate at 41.89%, compared with 7.40% among customers without internet service — a 34.49 percentage-point difference. This indicates that fiber-optic customers represent a particularly high-risk segment requiring further investigation.
